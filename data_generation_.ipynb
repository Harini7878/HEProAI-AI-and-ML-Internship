```python
import pandas as pd
import numpy as np

# Setting seed for consistent data generation
np.random.seed(42)
rows = 60

data = {
    'student_id': [f'S{1001+i}' for i in range(rows)],
    'age': np.random.randint(18, 25, size=rows),
    'program': np.random.choice(['B.Tech', 'MBA', 'M.Sc', 'B.A'], rows),
    'semester': np.random.randint(1, 9, size=rows),
    'gpa': np.round(np.random.uniform(5.0, 9.8, size=rows), 1),
    'attendance': np.random.randint(60, 100, size=rows),
    'assignments_completion': np.random.randint(40, 100, size=rows),
    'stress_level': np.random.randint(1, 11, size=rows),
    'sleep_hours': np.random.randint(4, 10, size=rows),
    'mental_wellbeing': np.random.randint(1, 11, size=rows),
    'productivity_score': np.random.randint(1, 11, size=rows),
    'distractions': np.random.randint(1, 11, size=rows),
    'career_clarity': np.random.randint(1, 11, size=rows),
    'skill_readiness': np.random.randint(1, 11, size=rows),
    'engagement_score': np.random.randint(30, 100, size=rows)
}

df = pd.DataFrame(data)

# Injecting Specific Behavioral Patterns for Activity 1
# 1. High stress (8+) + low productivity (<4)
df.loc[0:5, ['stress_level', 'productivity_score', 'mental_wellbeing']] = [9, 2, 3]

# 2. Low GPA (<6.0) + high engagement (>85)
df.loc[10:15, ['gpa', 'engagement_score', 'attendance']] = [5.5, 92, 95]

# 3. Strong academics (GPA > 9.0) + unclear career goals (<3)
df.loc[20:25, ['gpa', 'career_clarity', 'skill_readiness']] = [9.4, 2, 3]

df.to_csv('students.csv', index=False)
print("students.csv successfully generated .")
```

```python
import pandas as pd

# 1. Load your dataset from Activity 1
df = pd.read_csv('/content/students.csv')

# 2. Compute Domain Scores (Normalized to 100)
df['APS'] = (df['gpa'] * 10 + df['attendance']) / 2
df['WWS'] = ( (10 - df['stress_level']) * 10 + df['mental_wellbeing'] * 10 + df['sleep_hours'] * 10 ) / 3
df['PTMS'] = (df['assignments_completion'] + (df['productivity_score'] * 10)) / 2
df['CRS'] = (df['career_clarity'] * 10 + df['skill_readiness'] * 10) / 2

# 3. Compute overall Student Readiness Index (SRI)
df['SRI'] = df[['APS', 'WWS', 'PTMS', 'CRS']].mean(axis=1)

# 4. Classification Logic (Rule-Based Intelligence)
def classify_student(sri):
    if sri >= 85: return 'Blue (Excellent)'
    elif sri >= 70: return 'Green (Stable)'
    elif sri >= 50: return 'Yellow (Needs Attention)'
    else: return 'Red (At Risk)'

df['Category'] = df['SRI'].apply(classify_student)

# 5. Save the updated dataset
df.to_csv('student_scores.csv', index=False)
print("student_scores.csv generated.")

```
